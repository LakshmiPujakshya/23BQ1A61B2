```python
from getpass import getpass
import os

os.environ["GEMINI_API_KEY"] = getpass("Enter your Gemini API key: ")
```

```python
from google import genai
import os

# Initialize the client
client = genai.Client(api_key=os.environ["GEMINI_API_KEY"])

model_name = "gemini-3.8-flash"

print("[INFO] Starting Prompt Engineering Experiment...")
print("-" * 60)


# ==========================================
# EXPERIMENT PART 1: NAÏVE BASELINE PROMPT
# ==========================================

naive_prompt = "Write a one-paragraph bio of Ada Lovelace."

print("Running Baseline (Naïve) Prompt...")

naive_response = client.interactions.create(
    model=model_name,
    input=naive_prompt
)

naive_output = naive_response.output_text


# ==========================================
# EXPERIMENT PART 2: ENHANCED PROMPT
# ==========================================

enhanced_prompt = """
Role: You are an expert Science and Technology Historian writing an encyclopedia entry.

Task: Write a comprehensive, one-paragraph biographical summary of Ada Lovelace.

Specific Details to Include:
- Her mathematical collaboration with Charles Babbage.
- Her work on the Analytical Engine and why she is considered the first programmer.
- Her lineage as the daughter of Lord Byron.
- How her background blended arts and sciences.

Format Instructions:
- Output exactly ONE cohesive paragraph.
- Maximum 150 words.
- Tone: Scholarly yet highly engaging.
- Start with a powerful hook sentence highlighting her legacy.
"""

print("Running Enhanced Prompt...")

enhanced_response = client.interactions.create(
    model=model_name,
    input=enhanced_prompt
)

enhanced_output = enhanced_response.output_text


# ==========================================
# EXPERIMENT PART 3: EVALUATION
# ==========================================

print("\n" + "=" * 60)
print("EXPERIMENT RESULTS & COMPARISON")
print("=" * 60)

print("\n[OUTPUT A: BASELINE PROMPT]")
print(f'Prompt: "{naive_prompt}"')
print(f"Word Count: {len(naive_output.split())}")
print("-" * 40)
print(naive_output)
print("-" * 40)


print("\n[OUTPUT B: ENHANCED PROMPT]")
print("Prompt Features Used:")
print("Role Framing, Target Details, Structural Constraints")
print(f"Word Count: {len(enhanced_output.split())}")
print("-" * 40)
print(enhanced_output)
print("-" * 40)


# ==========================================
# CRITICAL ANALYSIS
# ==========================================

print("\nCRITICAL ANALYSIS & OBSERVATIONS")

print("\n1. RELEVANCE & SPECIFICITY:")
print("- Baseline: Provides a general biography.")
print("- Enhanced: Focuses on Babbage, Byron, the Analytical Engine, and programming.")

print("\n2. COMPLETENESS:")
print("- Baseline: May omit some requested details.")
print("- Enhanced: Explicitly includes the required information.")

print("\n3. STYLE & FORMAT ADHERENCE:")
print("- Baseline: Uses a general informational style.")
print("- Enhanced: Follows the historian persona, one-paragraph format, word limit, and requested tone.")
```
